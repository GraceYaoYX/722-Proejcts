```python
# a)
def get_late_percentage(promise, zone='all', time_block='all'):
    # Determine the zones to query
    zones_to_query = ZONES if zone == 'all' else [zone]

    # Determine the time blocks to query
    blocks_to_query = TIME_BLOCKS if time_block == 'all' else [time_block]

    all_delivery_times = []

    # Fetch and accumulate delivery times for each combination
    for z in zones_to_query:
        for tb in blocks_to_query:
            # Get times using the provided delivery_times function
            times = delivery_times(z, tb, promise)
            if len(times) > 0:
                all_delivery_times.extend(times)

    if not all_delivery_times:
        return 0.0

    all_delivery_times = np.array(all_delivery_times)

    # Calculate late percentage: delivery time strictly greater than promise
    late_orders = np.sum(all_delivery_times > promise)
    return (late_orders / len(all_delivery_times)) * 100

# Test the function with some examples
overall_late = get_late_percentage(45, 'all', 'all')
print(f"Overall late orders (45 min promise): {overall_late:.2f}%")
```

```python
all_late = get_late_percentage(45,'all','Lunch')
print(f"Overall late orders with 45 min promise: {all_late:.2f}%")
```

```python
# b)
results = []

for z in ZONES:
    for tb in TIME_BLOCKS:
        late_rate = get_late_percentage(45, z, tb)
        results.append(((z, tb), late_rate))

# Sort in descending order of late arrival rate
results.sort(key=lambda x: x[1], reverse=True)

print("Ranking of (Zone, Time Block) pairs by late arrival rate (descending):\n")
for idx, ((z, tb), rate) in enumerate(results, 1):
    print(f"{idx}. {z} - {tb}: {rate:.2f}%")
```

```python
# c)
def get_average_delivery_time(promise, zone='all', time_block='all'):
    # Determine the zones to query
    zones_to_query = ZONES if zone == 'all' else [zone]

    # Determine the time blocks to query
    blocks_to_query = TIME_BLOCKS if time_block == 'all' else [time_block]

    all_delivery_times = []

    # Fetch and accumulate delivery times for each combination
    for z in zones_to_query:
        for tb in blocks_to_query:
            # Get times using the provided delivery_times function
            times = delivery_times(z, tb, promise)
            if len(times) > 0:
                all_delivery_times.extend(times)

    if not all_delivery_times:
        return 0.0

    all_delivery_times = np.array(all_delivery_times)

    # Return the average delivery time
    return all_delivery_times.mean()
```

## Part 2

```python
# a)
def calculate_late_cost(COSTS):
    # Calculate the total cost of one late order
    return (COSTS["refund"]
        + COSTS["churn_orders"] * COSTS["margin"]
    )

# Test the function
print(f"Total cost per late order: ${calculate_late_cost(COSTS):.2f}")
```

```python
# b)
def best_promise(zone, time_block, promises, costs):
    """Return the promise with the highest net profit and its profit."""

    if len(promises) == 0:
        raise ValueError("The list of promises cannot be empty.")

    margin = costs["margin"]
    late_cost = calculate_late_cost(COSTS)

    recommended_promise = None
    highest_profit = float("-inf")

    for promise in sorted(promises):
        # Generate new data for each candidate promise
        times = delivery_times(zone, time_block, promise, seed=1)

        number_of_orders = len(times)
        number_of_late_orders = np.sum(times > promise)

        net_profit = (
            number_of_orders * margin
            - number_of_late_orders * late_cost
        )

        if net_profit > highest_profit:
            highest_profit = net_profit
            recommended_promise = promise

    return recommended_promise, highest_profit
```

```python
# Try promises from 5 to 90 minutes, in steps of 5
promises = list(range(5, 91, 5))

recommended, profit = best_promise(
    "Far West",
    "Fri/Sat eve",
    promises,
    COSTS
)

print(f"Recommended promise: {recommended} minutes")
print(f"Net profit over four weeks: ${profit:,.2f}")
```

To choose the proper range for the function, I initially consider the broad range included the promised delivery time of 45 minutes such as 5 - 90 minutes. Ideally, profit would reach its maximum inside this range. If the output shows the maximum profit appears at the boundary, we would expand the range as the profit may be still increasing near the boundary. The range needs to be custumized for each zone and time block as their delivery conditions may differ.
